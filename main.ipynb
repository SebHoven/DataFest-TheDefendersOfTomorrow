```python
import numpy as np
import pandas as pd
from matplotlib import pyplot
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import seaborn
```

```python
df = pd.read_csv("data/1.1 Totaal emissies - Nederland.csv", sep=";")
```

```python
df.columns

df[["gebied / dimensie"]]
```

```python
df["gebied / dimensie"] = df["gebied / dimensie"].str.replace("1.1 Totaal emissies    ", "").astype(str)

df["gebied / dimensie"] = df["gebied / dimensie"].replace("x", np.nan)

df["gebied / dimensie"] = df["gebied / dimensie"].str.replace(",", ".").astype(float)

df["gebied / dimensie"] = df["gebied / dimensie"].ffill()
```

```python
x = df[["Eenheid"]]
y = df["gebied / dimensie"]

model = LinearRegression()
model.fit(x, y)

toekomst_2030 = model.predict([[2030]])
toekomst_2050 = model.predict([[2050]])

print("toekomst 2030: ", toekomst_2030[0])
print("toekomst 2050: ", toekomst_2050[0])
print("1990: 167700")
print("verschil in percentage (1990/2030): ", (1 - (toekomst_2030[0] / 167700)) * 100)
print("verschil in percentage (1990/2050): ", (1 - (toekomst_2050[0] / 167700)) * 100)
```
